### Try to calculate the correlation and the corresponding p-value from our worked out class example

```python
# let's import the necessary libraries
import pandas as pd # this is same as nicknaming pandas as pd ot ease of use
from scipy import stats
```

```python
# Let's try using pandas to calculate the correlation
study = pd.DataFrame({'hours': [1, 2, 3, 4, 5, 6], 'score':[55, 50, 70, 65, 80, 85]})
print(f"The pearson correlation value (r) is: {study['hours'].corr(study['score'])}") # using pandas to calculate correlation (r)
```

Let's use stats from scipy to calculate the `pearson correlation (r)` as well as the `p-value`

```python
r, p = stats.pearsonr(study['hours'], study['score'])
print(f"The pearson correlation value (r) for hours practiced and scores obtained is: {r:.3f}")
print(f"The p-value for this correlation is: {p:.3f}")
```
