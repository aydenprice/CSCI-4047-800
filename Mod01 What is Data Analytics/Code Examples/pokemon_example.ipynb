```python
import pandas as pd

# I want all pokemon from the CSV to a data frame

df = pd.read_csv('AllPokemon.csv')
print(df)

# Displaying unique Names in pokemon

print(df['Name'].nunique(), "unique values in", len(df), "rows")

# display the number of Type1 unique values
print(list(df['Name'].unique()))




```

```python
print(list(df['Type2'].unique())) # shows all unique Type 2 occurrences 
```

```python
print(df.shape) # shows total # of rows 
```

```python
print(df.info()) # Lists all of the columns as well as the data types 
```
