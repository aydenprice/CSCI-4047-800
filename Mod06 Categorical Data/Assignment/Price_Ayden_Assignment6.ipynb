To measure the psychological state of Americans, a psychology researcher (in 1929) devised
three questions put to 3443 Americans from different walks of life. He asked them whether they
were happy. They could answer only **yes** or **no** to each question. By today’s standards the study
is quite simple, and he did no formal statistical analysis of his data. The frequency data (number
of **yes** and **no** responses within each employment category) from this study are in the file
*Americandata.csv*.

---

You need to use the data and answer the question, that is, **Is the type of profession related to the happiness of American people?**
Report the **χ2** statistic and the *statistical significance of test*, in addition to the contingency table.
Also, explain the results and what they mean. What does the value of **χ2** statistic mean? Is it
good or bad? Is the test statistically significant?

```python
#code here
import pandas as pd
import scipy.stats as stats

# Load the contingency table from the CSV file
df = pd.read_csv("Americandata.csv")

print(df)

wait = input("")

# Set Happy as the index to create the contingency table
contingency_table = df.set_index("Happy")

print("\nContingency Table:")
print(contingency_table)

wait = input("")

# Perform Chi-square test
chi2, p_value, dof, expected = stats.chi2_contingency(
    contingency_table,
    correction=False
)

# Create a DataFrame for expected frequencies
expected_df = pd.DataFrame(
    expected,
    index=contingency_table.index,
    columns=contingency_table.columns
)

print("\nExpected frequencies:")
print(expected_df.round(2))

# Check assumption
if (expected >= 5).all():
    print("\nAssumption Okay; all expected counts are greater than 5.")
else:
    print("\nWARNING! Some expected counts may be less than 5.")

# Report the numbers
print(f"\nChi-square = {chi2:.2f}")
print(f"Degrees of freedom = {dof}")
print(f"p-value = {p_value:.2e}")

# Significance level
alpha = 0.05

# Find critical value
critical_value = stats.chi2.ppf(1 - alpha, dof)

print(
    f"Critical value (alpha = {alpha} and degree of freedom = {dof}) "
    f"is: {critical_value:.2f}"
)

# Decision using critical value
if chi2 > critical_value:
    print(
        f"Since {chi2:.2f} > {critical_value:.2f}, "
        "Reject the null hypothesis -> profession and happiness ARE related"
    )
else:
    print(
        f"Since {chi2:.2f} <= {critical_value:.2f}, "
        "Cannot reject the null hypothesis"
    )

# Decision using p-value
if p_value < alpha:
    print(
        f"p < {alpha}: REJECT the null hypothesis "
        "-> profession and happiness ARE related"
    )
else:
    print(
        f"p >= {alpha}: Cannot reject the null hypothesis "
        "-> no evidence that profession and happiness are related"
    )
```

The chi-square test gave χ^2 = 936.14, df = 7, and p = 0. This is way above the critical value of 14.07, so we reject the null hypothesis: profession and happiness ARE related. A chi square this large means a huge difference between what we saw and what we'd expect if profession didn't matter, so it's a strong, significant result, not random chance. For example, Laborers and Physicians were happy 75%+ of the time, while Students and Lawyers were happy less than 20% of the time.
